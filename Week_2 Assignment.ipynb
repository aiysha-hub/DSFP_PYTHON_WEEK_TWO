```python
#Write a function that counts vowels in a string.

def count_vowels(text):
    vowels = "aeiouAEIOU"  # include uppercase letters too
    count = 0
    for char in text:
        if char in vowels:
            count += 1
    return count


print(count_vowels("Hello World"))       # 3
print(count_vowels("I am beautiful"))    # 6
print(count_vowels("AEIOU aiou eio aou"))       # 15
```

```python
#Write a function that checks if a word is a palindrome.

def is_palindrome(word):
    word = word.lower()        # make it case-insensitive
    return word == word[::-1]  # compare with reversed word

print(is_palindrome("madam"))   # True
print(is_palindrome("Racecar")) # True
print(is_palindrome("python"))  # False
print(is_palindrome("level"))   #True
print(is_palindrome("hello"))   #False
```

```python
# Q14. Write a function with a default parameter for tax rate.

def tax(individual_name, individual_income, tax_rate=0.1):
    tax_payable = individual_income * tax_rate

    if tax_payable <= 10000:
        category = "Don't pay tax"
    else:
        category = "Pay tax"

    result = {
        "individual_name": individual_name,
        "individual_income": individual_income,
        "tax_rate": tax_rate,
        "tax_payable": tax_payable,
        "category": category
    }

    return result

print(tax("Aysha", 900000))        # uses default tax_rate = 0.1
print(tax("Aysha", 50000, 0.15))   # overrides the default
```

```python
# Q27. Write a function that checks if a number is prime.

def is_prime(num):
    # Numbers less than 2 are not prime
    if num < 2:
        return False
    # Check divisibility from 2 up to the square root of n
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

# Example usage
print(is_prime(17))   # True
print(is_prime(28))  # False
```

```python
# Q28. Write a function that returns the sum of digits of a number.

def sum_of_digits(n):
    n = abs(n)  # Make sure it works for negative numbers too
    total = 0
    while n > 0:
        total += n % 10  # Add the last digit
        n = n // 10      # Remove the last digit
    return total

# Example usage
print(sum_of_digits(1234))  # 10 (1+2+3+4)
print(sum_of_digits(567))  # 18 (5+6+7)
```

```python
# Q29. Write a function that accepts a list of prices and applies a default discount.

def default_discount(prices, discount=10):
    return [price * (1 - discount/100) for price in prices]

# Example usage with a list of prices
print(default_discount([100, 578]))
```
