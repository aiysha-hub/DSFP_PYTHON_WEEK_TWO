```python
# This function is a way to accept any number of inputs when calling a function.
#It allows a function to accept any number of inputs without breaking.
#*args collects extra unnamed values into a tuple, and **kwargs collects extra named values into a dictionary.
#The idea behind this function is to make a function that can accept any amount of information and inspect or handle it later.

def get_info(*args, **kwargs):
    print(args)
    print(kwargs)
```

```python
get_info("Mamah", "Kemi", 67, age = 23, city="Kaduna", job="Analyst")
```

```python
#create a function that can let a user multiply or divide two numbers a, b.

def calc(x, y, op):
    if op == '*':
        return x * y
    elif op == '/':
        return x / y 
```

```python
x = 9
y = 3
op = '*'
calc(x, y, op)
```

```python
x =20
y = 4
op = '/'
calc(x, y, op)
```

```python
print (calc (5, 4, "*"))
```

```python
print (calc (50, 10, '/'))
```
